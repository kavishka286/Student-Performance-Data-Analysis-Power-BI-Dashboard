```python
# Create sample student performance data

np.random.seed(42)

students = 100

student_data = {
    "Student_ID": [f"S{i:03d}" for i in range(1, students + 1)],
    "Gender": np.random.choice(["Male", "Female"], students),
    "Study_Hours": np.random.randint(4, 16, students),
    "Attendance": np.random.randint(70, 101, students),
    "Previous_Score": np.random.randint(50, 91, students),
    "Assignment_Score": np.random.randint(50, 101, students),
    "Sleep_Hours": np.random.randint(5, 10, students)
}

df = pd.DataFrame(student_data)

# Calculate Final Score
df["Final_Score"] = (
    df["Study_Hours"] * 1.5
    + df["Attendance"] * 0.2
    + df["Previous_Score"] * 0.25
    + df["Assignment_Score"] * 0.25
    + df["Sleep_Hours"] * 1
)

# Keep scores between 0 and 100
df["Final_Score"] = df["Final_Score"].clip(0, 100).round(0).astype(int)

# Save the dataset
df.to_csv("../data/student_performance.csv", index=False)

print("Dataset created successfully!")
print(f"Number of students: {len(df)}")
```
